```python
import os
import pandas as pd
from IPython.display import display  # 专门用来交互显示

def read_china_sheets(folder_path):
    all_data = {}
    for file in os.listdir(folder_path):
        if file.endswith(".xlsx") or file.endswith(".xls"):
            file_path = os.path.join(folder_path, file)
            try:
                df = pd.read_excel(file_path, sheet_name="CHINA", header=1) 
                all_data[file] = df
                print(f"✅ 成功读取 {file} 中的 'CHINA' sheet")
            except Exception as e:
                print(f"⚠️ 跳过 {file}，原因: {e}")
    return all_data


if __name__ == "__main__":
    folder = r"D:\UN\02-UN DESA world urbanization prospects"
    china_sheets = read_china_sheets(folder)

    # 交互式展示所有文件
    if china_sheets:
        for file, df in china_sheets.items():
            print(f"\n📂 文件名: {file}")
            display(df.head())  # 交互式展示（带表格格式）
    else:
        print("❌ 没有读取到任何数据")
```
